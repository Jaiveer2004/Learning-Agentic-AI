# Tokenization

Tokenization means breaking text into smaller pieces called **tokens**. A token can be a sentence, a word, or even punctuation.

Computers cannot directly understand a complete paragraph like humans do. Tokenization gives the text a clear structure so NLP models can process it.

Typical steps are:
1. **Sentence tokenization:** paragraph → sentences.
2. **Word tokenization:** sentence or paragraph → words and punctuation.

The notebook uses NLTK, a Python library with tools for working with human language.

## Example corpus

A corpus is a collection of text. Here, we create a small paragraph and use it to demonstrate the different tokenization methods.

```python
corpus = """Hello Welcome, to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

## 1. Sentence tokenization

`sent_tokenize` separates a paragraph into complete sentences.

```python
from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
print(documents)
```

## 2. Word tokenization

`word_tokenize` separates text into words and punctuation. It can be applied to the complete corpus or to each sentence.

```python
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

## 3. Word-and-punctuation tokenization

`wordpunct_tokenize` uses simple boundaries between words and punctuation. Compare its treatment of `Naik's` with `word_tokenize`.

## Notes for quick understanding

### `sent_tokenize`
- Splits a paragraph into complete sentences.
- It looks at punctuation and language rules to identify sentence boundaries.
- In the example, the paragraph becomes three separate sentences.

### `word_tokenize`
- Splits text into words and punctuation marks.
- Punctuation is kept as a separate token, so `.`, `,`, and `!` also appear in the output.
- It treats contractions carefully. For example, `Naik's` may become `Naik` and `'s`.

### `wordpunct_tokenize`
- Splits at word and punctuation boundaries using a simpler rule.
- It may split contractions into more separate pieces, such as `'` and `s`.
- It is simple and useful when punctuation must be separated clearly, but `word_tokenize` is often more language-aware.

### Why tokenization matters
Tokenization is usually one of the first NLP steps. After text is split into tokens, we can count words, remove unwanted tokens, create numerical features, and pass the text to a machine-learning model.

## TLDR (All in one for one shot revision, crisp and short)

Tokenization breaks text into smaller tokens. `sent_tokenize` creates sentences, `word_tokenize` creates language-aware words and punctuation, and `wordpunct_tokenize` uses simpler word/punctuation boundaries. It is a basic first step before NLP models can process text.
