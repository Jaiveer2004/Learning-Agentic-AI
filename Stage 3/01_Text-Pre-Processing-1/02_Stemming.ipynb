# Stemming

Stemming means reducing a word to a shorter, root-like form.

Example: `playing`, `played`, and `plays` may be reduced to `play`.

This helps NLP systems treat different forms of a word as related during search, classification, and sentiment analysis.

**Important:** A stem is not always a real English word. Rule-based stemming can produce forms such as `histori` from `history`.

A **lemma** is the correct dictionary base form of a word. Stemming usually removes word endings, while lemmatization uses vocabulary and grammar to find the correct word.

## Classification Problem
### Classify whether a product review is positive or negative.

Words with similar meanings may appear in different forms. Stemming can reduce some of these forms so the model can focus on the main word.

Examples:
1. `eating`, `eat`, `eaten` may be reduced toward `eat`.
2. `going`, `gone`, `went` are related in meaning, but a basic stemmer may not convert all of them to `go` because they are irregular forms.

## Example words

We will apply different stemmers to the same list of words so that their outputs can be compared.

```python
words = ["eating", "eat", "eaten", "going", "gone", "went", "programming", "programmer", "programmed", "history", "historical", "historically", "finally", "final", "finalize", "finalized"]
```

## 1. PorterStemmer

PorterStemmer is a classic, fast, rule-based English stemmer. It removes common endings, but the result may not be a complete English word.

```python
from nltk.stem import PorterStemmer

stemming = PorterStemmer()
for word in words:
    print(word + ': ' + stemming.stem(word))
    
# `histori` is a stem, not a correct English word. This is a limitation of stemming.
```

```python
stemming.stem('congratulations')
```

## 2. RegexStemmer Class

`RegexpStemmer` lets us define our own rule with a regular expression. It removes a matching prefix or suffix from a word.

In this example, `ing$|s$|e$|able$` means: remove `ing`, `s`, `e`, or `able` when it appears at the end of a word. The `$` symbol means "end of the word".

`min=4` prevents the stemmer from making a word shorter than four characters. This gives us control, but the result depends completely on the rules we write.

### RegexStemmer examples

The examples below show how the custom rule behaves for normal words, a word with an extra prefix, and a plural word.

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min = 4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer.stem('congratulations')
```

## 3. SnowballStemmer

SnowballStemmer is an improved rule-based stemmer that supports several languages. It is often a better choice than PorterStemmer, but it can still produce a non-word.

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word + ": " + snowball_stemmer.stem(word))
    
# `histori` is still not a real word, but Snowball usually handles more patterns than PorterStemmer.
```

```python
stemming.stem('fairly'), stemming.stem('sportingly')
```

```python
snowball_stemmer.stem('fairly'), snowball_stemmer.stem('sportingly')
```

### Comparing `fairly` and `sportingly`

SnowballStemmer reduces both words to more natural stems than PorterStemmer in this example. This shows why testing a stemmer on representative words is important.

## Notes for quick understanding

### What stemming does
- It removes common beginnings or endings from words.
- It is fast and useful when similar word forms should be grouped together.
- It does not understand the full meaning of a sentence.
- It can create an incomplete or incorrect-looking word, such as `histori`.

### Comparing the stemmers
- **PorterStemmer:** A classic English stemmer. It is simple and fast, but its output may not be a real word.
- **RegexpStemmer:** Uses rules that we write ourselves. It is easy to understand and customize, but poor rules can give poor results.
- **SnowballStemmer:** A newer and generally improved rule-based stemmer. It supports multiple languages, but it can still produce non-words such as `histori`.

### Stemming vs. lemmatization
- **Stemming:** Fast rule-based shortening; output may not be a dictionary word.
- **Lemmatization:** Finds the correct dictionary form using vocabulary and language information; usually more accurate but slower and dependent on language resources.

### When to use it
Use stemming when speed is important and an approximate root is enough, such as basic search or some text-classification tasks. Use lemmatization when readable and linguistically correct words matter.

## TLDR (All in one for one shot revision, crisp and short)

Stemming reduces words to shorter root-like forms for NLP. Porter is simple, Regex is fully custom, and Snowball is usually better than Porter. Stems may not be real words (`history` → `histori`). Stemming is fast but approximate; lemmatization is slower but returns correct dictionary forms.
