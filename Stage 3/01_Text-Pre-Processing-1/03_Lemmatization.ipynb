# Lemmatization

Lemmatization reduces a word to its correct dictionary form, called a **lemma**.

Examples:
- `running` → `run`
- `better` → `good` when the word is used as an adjective
- `children` → `child`

Unlike stemming, lemmatization tries to return a real word with the same basic meaning. It uses vocabulary and language information, so it is usually more accurate but can be slower.

## Lemmatization vs. stemming

| Technique | How it works | Example | Main point |
|---|---|---|---|
| Stemming | Removes word endings using rules | `history` → `histori` | Fast, but the result may not be a real word |
| Lemmatization | Uses a dictionary and grammar information | `children` → `child` | More meaningful, but usually slower |

The correct result can depend on the word's **part of speech (POS)**. For example, `better` can be related to `good` as an adjective, but it can have a different result when used as a verb.

## WordNetLemmatizer

NLTK provides `WordNetLemmatizer`, which uses the WordNet lexical database. WordNet contains information about English words and their relationships.

Run the next cell once to download the data required by the lemmatizer.

```python
import nltk

nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatizer.lemmatize("going")
```

## Why POS matters

By default, `lemmatize()` assumes that a word is a **noun**. This is why adding the correct POS tag often gives a better result.

WordNet uses these POS codes:
- `n` = noun
- `v` = verb
- `a` = adjective
- `r` = adverb

```python
examples = [
    ("cars", "n"),
    ("running", "v"),
    ("easily", "r"),
    ("better", "a"),
]

for word, pos in examples:
    print(f"{word} ({pos}): {lemmatizer.lemmatize(word, pos=pos)}")
```

```python
words = ["eating", "eat", "eaten", "going", "gone", "went", "programming", "programmer", "programmed", "history", "historical", "historically", "finally", "final", "finalize", "finalized"]
```

```python
for word in words:
    print(word + ": " + lemmatizer.lemmatize(word, pos = 'v'))
```

```python
lemmatizer.lemmatize('fairly', pos='v'), lemmatizer.lemmatize('sportingly')
```

## Notes for quick understanding

- Lemmatization returns a valid dictionary form in many cases.
- It is not the same as simply cutting off the last few letters.
- The context and POS of a word affect the answer.
- If the POS is not supplied, NLTK assumes the word is a noun.
- WordNet is mainly designed for English, so results depend on the available language resource.
- Lemmatization is useful before search, text classification, sentiment analysis, and other NLP tasks when meaningful word forms are important.
