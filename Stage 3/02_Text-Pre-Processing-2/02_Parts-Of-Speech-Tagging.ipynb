# Parts-of-Speech (POS) Tagging

POS tagging assigns a grammatical label to each word in a sentence. For example, a noun is labelled as a noun, a verb as a verb, and an adjective as an adjective.

POS tags help NLP systems understand how words are being used. The same word can have different roles depending on its context, so tagging is based on the complete sentence rather than only the word itself.

## Example and POS tag reference

"Taj Mahal is a beautiful Monument"

Common Penn Treebank POS tags:

| Tag | Meaning | Example |
|---|---|---|
| `CC` | coordinating conjunction | and |
| `CD` | cardinal number | three |
| `DT` | determiner | the, a |
| `IN` | preposition or subordinating conjunction | in, because |
| `JJ`, `JJR`, `JJS` | adjective, comparative, superlative | big, bigger, biggest |
| `MD` | modal verb | could, will |
| `NN`, `NNS` | singular or plural noun | desk, desks |
| `NNP`, `NNPS` | singular or plural proper noun | India, Americans |
| `PRP`, `PRP$` | personal or possessive pronoun | he, my |
| `RB`, `RBR`, `RBS` | adverb, comparative, superlative | quickly, better, best |
| `VB`, `VBD`, `VBG`, `VBN`, `VBP`, `VBZ` | verb forms | take, took, taking, taken, takes |
| `WDT`, `WP`, `WP$`, `WRB` | question words | which, who, whose, where |

The complete tag set contains additional labels. These common tags are enough for the examples in this notebook.

## 1. Prepare the text

We use a paragraph as the input corpus. Later, it is split into sentences and words before tagging.

```python
paragraph = """I have three visions for India. In 3000 years of our history, people from all over 
               the world have come and invaded us, captured our lands, conquered our minds. 
               From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British,
               the French, the Dutch, all of them came and looted us, took over what was ours. 
               Yet we have not done this to any other nation. We have not conquered anyone. 
               We have not grabbed their land, their culture, 
               their history and tried to enforce our way of life on them. 
               Why? Because we respect the freedom of others.That is why my 
               first vision is that of freedom. I believe that India got its first vision of 
               this in 1857, when we started the War of Independence. It is this freedom that
               we must protect and nurture and build on. If we are not free, no one will respect us.
               My second vision for India’s development. For fifty years we have been a developing nation.
               It is time we see ourselves as a developed nation. We are among the top 5 nations of the world
               in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling.
               Our achievements are being globally recognised today. Yet we lack the self-confidence to
               see ourselves as a developed nation, self-reliant and self-assured. Isn’t this incorrect?
               I have a third vision. India must stand up to the world. Because I believe that unless India 
               stands up to the world, no one will respect us. Only strength respects strength. We must be 
               strong not only as a military power but also as an economic power. Both must go hand-in-hand. 
               My good fortune was to have worked with three great minds. Dr. Vikram Sarabhai of the Dept. of 
               space, Professor Satish Dhawan, who succeeded him and Dr. Brahm Prakash, father of nuclear material.
               I was lucky to have worked with all three of them closely and consider this the great opportunity of my life. 
               I see four milestones in my career"""
```

## 2. Load NLTK and split the paragraph

The paragraph is split into sentences because POS tagging works on tokenized text.

```python
import nltk
```

```python

from nltk.corpus import stopwords
sentences=nltk.sent_tokenize(paragraph)
```

```python
sentences
```

## 3. Apply POS tagging

The NLTK POS tagger predicts one tag for each token. Stop words are removed here to keep the output smaller, but removing them is optional and depends on the task.

The tagger needs the averaged perceptron model downloaded in the next cell.

```python
nltk.download('averaged_perceptron_tagger')
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[word for word in words if word not in set(stopwords.words('english'))]
    #sentences[i]=' '.join(words)# converting all the list of words into sentences
    pos_tag=nltk.pos_tag(words)
    print(pos_tag)
```

## Notes for quick understanding

- POS tagging gives each token a grammatical label, such as noun (`NN`), verb (`VB`), adjective (`JJ`), or adverb (`RB`).
- The tag depends on context. For example, a word can have different roles in different sentences.
- Punctuation is also returned with a punctuation tag such as `.` or `,`.
- Proper nouns such as `India`, `Taj`, and `Mahal` are commonly tagged as `NNP`.
- POS tagging is useful for lemmatization, information extraction, grammar analysis, and many other NLP tasks.
- Taggers are not perfect. Names, contractions, unusual formatting, and domain-specific words can be tagged incorrectly.
