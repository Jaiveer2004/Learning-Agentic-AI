# Named Entity Recognition (NER)

Named Entity Recognition finds important real-world names in text and labels them by type.

For example, an NER system can identify a person's name, a place, an organization, a date, or an amount of money. This helps NLP applications understand **who**, **where**, **when**, and **what** a text is talking about.

## Common entity types

| Entity type | Meaning | Example |
|---|---|---|
| `PERSON` | Name of a person | Krish C Naik |
| `GPE` / `LOCATION` | Country, city, or place | India |
| `DATE` | A date or year | September, 24-09-1989 |
| `TIME` | A time | 4:30 pm |
| `MONEY` | A monetary amount | 1 million dollars |
| `ORGANIZATION` | A company or institution | iNeuron Private Limited |
| `PERCENT` | A percentage | 20%, twenty percent |

The exact labels depend on the NER model. NLTK's chunker commonly uses labels such as `PERSON`, `GPE`, `ORGANIZATION`, and `DATE`.

## Example sentence

We will use this sentence to find named entities:

> The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures.

## 1. Prepare the text

NER normally works after tokenization and POS tagging. First, the sentence is split into individual tokens.

```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

```python
import nltk
words=nltk.word_tokenize(sentence)
words
```

```python
tag_elements=nltk.pos_tag(words)
tag_elements
```

## 2. Download NLTK resources

`maxent_ne_chunker` contains the NLTK named-entity chunking model. The `words` corpus provides vocabulary data used by the chunker.

```python
nltk.download('maxent_ne_chunker')
```

```python
nltk.download('words')
```

## 3. Recognize named entities

`ne_chunk()` groups related tokens into named-entity phrases. We display the tree as text because NLTK's `.draw()` opens a Tkinter window and can fail in some VS Code or notebook environments.

```python
%pip install numpy svgling
```

```python
named_entities = nltk.ne_chunk(tag_elements)
named_entities
```

## Notes for quick understanding

- **Tokenization** breaks the sentence into words and punctuation.
- **POS tagging** identifies the grammatical role of each token.
- **NER** combines this information to identify names and other important entities.
- NER is different from POS tagging: `NNP` means a proper noun, while `PERSON` identifies that the proper noun is a person's name.
- Entity recognition is model-based, so unusual names, spelling, punctuation, and domain-specific text can produce incorrect labels.
- NER is useful for search, question answering, document analysis, news analysis, and extracting structured data from text.
- NLTK's `.draw()` opens a Tkinter visual tree and may fail in VS Code or notebook environments, so the example displays the tree text instead. For production applications, a library such as spaCy can provide easier programmatic access to entity spans and labels.

Named Entity Recognition finds and labels real-world entities such as people, places, dates, organizations, money, and percentages. The usual flow is tokenization → POS tagging → NER chunking. NER is useful for extracting structured information, but its predictions depend on the model and context.
