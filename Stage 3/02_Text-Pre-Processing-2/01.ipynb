# Stop Words, Stemming, and Lemmatization

This notebook shows a common NLP text-cleaning pipeline:

1. Split a paragraph into sentences and words.
2. Remove common words called **stop words**.
3. Reduce the remaining words with stemming or lemmatization.

Stop words often include words such as `the`, `is`, `and`, and `to`. They are common and may add little value for some tasks, but the correct list depends on the problem. Always check whether removing them changes the meaning you need.

## 1. Example paragraph

The paragraph below is the input text. We will apply the same cleaning idea to it using different word-reduction methods.

```python
paragraph = """I have three visions for India. In 3000 years of our history, people from all over 
               the world have come and invaded us, captured our lands, conquered our minds. 
               From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British,
               the French, the Dutch, all of them came and looted us, took over what was ours. 
               Yet we have not done this to any other nation. We have not conquered anyone. 
               We have not grabbed their land, their culture, 
               their history and tried to enforce our way of life on them. 
               Why? Because we respect the freedom of others.That is why my 
               first vision is that of freedom. I believe that India got its first vision of 
               this in 1857, when we started the War of Independence. It is this freedom that
               we must protect and nurture and build on. If we are not free, no one will respect us.
               My second vision for India's development. For fifty years we have been a developing nation.
               It is time we see ourselves as a developed nation. We are among the top 5 nations of the world
               in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling.
               Our achievements are being globally recognised today. Yet we lack the self-confidence to
               see ourselves as a developed nation, self-reliant and self-assured. Isn’t this incorrect?
               I have a third vision. India must stand up to the world. Because I believe that unless India 
               stands up to the world, no one will respect us. Only strength respects strength. We must be 
               strong not only as a military power but also as an economic power. Both must go hand-in-hand. 
               My good fortune was to have worked with three great minds. Dr. Vikram Sarabhai of the Dept. of 
               space, Professor Satish Dhawan, who succeeded him and Dr. Brahm Prakash, father of nuclear material.
               I was lucky to have worked with all three of them closely and consider this the great opportunity of my life. 
               I see four milestones in my career"""
```

## 2. Install and load NLTK resources

NLTK provides the stop-word list, tokenizers, stemmers, and lemmatizer used below. The download cell is needed only when these resources are not already installed.

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
stemmer = PorterStemmer()
```

## 3. Remove stop words and apply Porter stemming

First, each sentence is split into words. Words found in the English stop-word list are removed. PorterStemmer then shortens the remaining words.

The output may contain forms that are not real English words, such as `histori`. This is normal for stemming.

```python
sentences = nltk.sent_tokenize(paragraph)
```

```python
# Apply stopwords and filter and then apply stemming
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) #Converting all the words into sentences.
    
sentences
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer = SnowballStemmer('english')
```

## 4. Remove stop words and apply Snowball stemming

SnowballStemmer applies a more developed set of stemming rules than PorterStemmer. It is still rule-based, so its output may not always be a dictionary word.

```python

# Apply stopwords and filter and then apply Snowball stemming
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [snowball_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) #Converting all the words into sentences.
    
sentences
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

## 5. Remove stop words and apply lemmatization

Lemmatization tries to return a meaningful dictionary form. Here, `pos='v'` tells WordNet that the words should be treated as verbs. The correct part of speech can improve the result.

```python

# Apply stopwords and filter and then apply lemmatizer
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word, pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) #Converting all the words into sentences.
    
sentences
```

## Notes for quick understanding

### Stop words
- Stop words are very common words that may not help a particular NLP task.
- Removing them can reduce noise and the number of tokens.
- Do not remove them automatically when grammar, negation, or exact meaning matters. For example, removing `not` can reverse the meaning of a sentence.

### Stemming and lemmatization in this pipeline
- **PorterStemmer:** Fast and simple, but may create incomplete words.
- **SnowballStemmer:** Usually handles more patterns than Porter, but can also create non-words.
- **WordNetLemmatizer:** Tries to produce a valid dictionary form and works best when the correct POS is provided.

### Important notebook detail
Each loop changes the existing `sentences` list in place. To compare Porter, Snowball, and lemmatization fairly, rerun the sentence-tokenization cell before running each method, or create a fresh copy of the original sentences for every experiment.

## TLDR (All in one for one shot revision, crisp and short)

Stop-word removal deletes common words, then stemming or lemmatization reduces the remaining words. Porter is fast but rough, Snowball is usually better, and lemmatization gives more meaningful dictionary forms when POS is correct. Always check whether removing stop words changes the task's meaning.
